#step 2 :- Reduces noise first

```python
analysis_cols = [
    "Title",
    "Genre",
    "Series or Movie",
    "Runtime",
    "IMDb Score",
    "IMDb Votes",
    "Rotten Tomatoes Score",
    "Metacritic Score",
    "Release Date"
]

df_analysis = df[analysis_cols]
df_analysis = df[analysis_cols].copy()

```

```python
df_analysis.info()
```

```python
df_analysis.columns
```

```python
df_analysis.head()
```

```python
#data cleaning
```

```python
df_analysis["Runtime"]=(
    df_analysis["Runtime"]
    .str.extract(r"(\d+)")
    .astype("float")
)
```

```python
df_analysis.isnull().sum()
```

```python
df_analysis.loc[:, "IMDb Votes"] = (
    df_analysis["IMDb Votes"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)
df_analysis = df_analysis[df_analysis["IMDb Votes"] >= 1000]

```

```python
df_analysis["Runtime"].dtype
```

```python
#step 4 : start answring questions
```

```python
df_analysis["IMDb Score"].describe()
```

```python
import matplotlib.pyplot as plt
imdb_scores = df_analysis["IMDb Score"].dropna()
plt.hist(imdb_scores, bins=20)
plt.xlabel("IMDb Rating")
plt.ylabel("count")
plt.title("Distribution of IMDb Ratings for Netflix Titles")
plt.show()
```

```python
rating_corr = df_analysis[
    ["IMDb Score", "Rotten Tomatoes Score", "Metacritic Score"]
].corr()

rating_corr
```

```python
#what company cares about
```

```python
df_genre = df_analysis.copy()
df_genre["Genre"] = df_genre["Genre"].str.split(",")
df_genre = df_genre.explode("Genre")
df_genre["Genre"] = df_genre["Genre"].str.strip()
```

```python
genre_rating = (
    df_genre
    .groupby("Genre")["IMDb Score"]
    .mean()
    .sort_values(ascending=False)

)
genre_rating.head(10)
```

```python
genre_rating.head(10).plot(kind="barh", color="skyblue", figsize=(10, 6))
plt.xlabel("Average IMDb Score")
plt.ylabel("Genre")
plt.title("Top 10 IMDb Score by Genre")
plt.gca().invert_yaxis()  # highest rating on top
plt.show()
```

```python
#compare  netflix rating on imdb vs rotten tomatoes vs metacritic
```

```python
df_compare = df_analysis.copy()
df_compare["IMDb Score (100)"] = df_compare["IMDb Score"] * 10
```

```python
df_compare = df_compare.dropna(
    subset=["IMDb Score (100)", "Rotten Tomatoes Score", "Metacritic Score"],
    how="all"
)
```

```python
corr_matrix = df_compare[
    ["IMDb Score (100)", "Rotten Tomatoes Score", "Metacritic Score"]
].corr()
corr_matrix
```

```python
plt.scatter(
    df_compare["IMDb Score (100)"],
    df_compare["Rotten Tomatoes Score"],
    alpha=0.4
)
plt.xlabel("IMDb Score (Audience)")
plt.ylabel("Rotten Tomatoes (Score)")
plt.title("Audience vs Critics: IMDb Score vs Rotten Tomatoes Score")
plt.show()
```

```python
plt.scatter(
    df_compare["IMDb Score (100)"],
    df_compare["Metacritic Score"],
    alpha=0.4
)
plt.xlabel("IMDb Score (Audience)")
plt.ylabel("Metacritic Score (Critics)")
plt.title("Audience vs Critics: IMDb vs Metacritic")
plt.show()
```
